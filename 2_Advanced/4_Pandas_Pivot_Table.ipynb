```python
#import library
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()

#data cleaning
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.pivot_table(index='job_title_short', values='salary_year_avg', columns='job_country', aggfunc='median')
```

```python
topcountries = df['job_country'].value_counts().head(6).index
```

```python
df_job_country_salary = df.pivot_table(
    index='job_country', 
    values='salary_year_avg', 
    columns='job_title_short', 
    aggfunc='median')

df_job_country_salary = df_job_country_salary.loc[topcountries]

job_titles=['Data Analyst', 'Data Engineer', 'Data Scientist']

df_job_country_salary = df_job_country_salary[job_titles]

df_job_country_salary.plot(
    kind='bar', 
    rot=45, 
    xlabel="", 
    ylabel="Median Salary($USD)", 
    title="Median Salary by Job Title and Country")
plt.show()
```
