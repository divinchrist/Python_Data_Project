#Loops
For

While

```python
numbers=[0,1,2,3,4,5]
for number in numbers:
  print(number)
```

```python
skill='python'
for character in skill:
  print(character)
```

```python
year_of_experience ={
    'luke': 3,
    'kelly': 6,
    'ken': 4,
    'Alex': 3
}
```

```python
for applicant in year_of_experience:
  print(applicant)
```

```python
for value in year_of_experience.values():
  print(value)
```

```python
for items in year_of_experience.items():
  print(items)
```

```python
for person, value in year_of_experience.items():
  if value >3:
    print(f'{person} is experienced with {value}')
```

```python
for i in  range(5):
  print(i)
```

#While Loops

```python
i=1
while i<5:
    print(i, 'is less than 5')
    i+=1
```
