#String Formatting

```python
role= "Data Analys"
skill= "Python"
```

```python
print("role: "+ role)
```

```python
print("Role: {}".format(role))
```

```python
print ("Role: {}, Skill: {}".format(role,skill))
```

```python
print(f"Role: {role}; Skill: {skill}")
```

```python
print("Role: %s" %role)
```

```python
YearOfExperience= 123456789

print(",".join(YearOfExperience))
```
