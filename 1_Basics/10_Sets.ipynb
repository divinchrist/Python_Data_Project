#Sets
Generally used to make sure we have unique value in a list for example: see the last lines of code

```python
job_skills= {'tableau',  'sql', 'python','statistics'}
print(job_skills)
```

```python
job_skills.add('excel')
print(job_skills)
```

```python
job_skills.add('sql')
print(job_skills)
```

```python
print(job_skills.pop())
```

```python
job_skills.remove('tableau')
print(job_skills)
```

```python
skill_list=['python', 'sql', 'statistics', 'tableau', 'python', 'sql', 'statistis', 'tableau']
```

```python
list(set(skill_list))
```
