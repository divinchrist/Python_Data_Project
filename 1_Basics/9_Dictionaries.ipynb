#Dictionaries

```python
job_type_skills={'database': 'postgresql',
 'language':'python',
 'library': 'pandas'}
```

```python
print(job_type_skills)
```

```python
print(job_type_skills['database'])
```

```python
print(job_type_skills.keys())
```

```python
print(job_type_skills.values())
```

```python
job_type_skills.pop('library')
print(job_type_skills)
```

```python
job_type_skills.update({'cloud':'google cloud'})
```

```python
print (job_type_skills)
```

```python
print(job_type_skills['cloud'])
```

```python
job_type_skills[{'version_control'}]='git'
print(job_type_skills)
```
